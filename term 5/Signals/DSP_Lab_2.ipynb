### 1. Непрерывные системы

####  1.1. Дифференциальное уравнение системы

Для непрерывной линейной стационарной системы связь между входным $x(t)$ и выходным $y(t)$ сигналами может быть задана линейным обыкновенным дифференциальным уравнением n-го порядка:

$$
a_n\frac{d^ny(t)}{dt^n}+a_{n-1}\frac{d^{n-1}y(t)}{dt^{n-1}}+\ldots
+a_1\frac{dy(t)}{dt}+a_0y(t)
= b_m\frac{d^mx(t)}{dt^m}+\ldots+b_1\frac{dx(t)}{dt}+b_0x(t) \tag{1}
$$


_В большинстве приложений вы скорее всего сталкивались с редуцированной формой:_
$$
\sum_{i=0}^{n}a_i\frac{d^iy(t)}{dt^i}
= x(t) \tag{1*}
$$
Перейти от (1) к системе дифференциальных уравнений первого порядка можно с помощью введения дополнительных переменных:
$$\xi_i = \frac{d^iy(t)}{dt^i}, \space\space i = 0:n-1$$

### Преобразование Лапласа
Интегральное преобразование Лапласа для непрерывной функции $f(t)$:

$$ F(s) = \mathcal{L}(f(t)) =  ∫_0^∞f(t)e^{-st}dt $$
Преобразование Лапласа может использоваться для перехода от дифференциального уравнения системы к алгебраическому уравнению в $s$-области. Обозначим за:

$$
X(s)=\mathcal{L}\{x(t)\},
\qquad
Y(s)=\mathcal{L}\{y(t)\}.
$$
и примем:
$$
\forall i\in[0,n] \space\space y^{(i)}(0) = 0, \space\space \forall j\in[0,m] \space\space x^{(j)}(0) = 0
$$

Тогда применив преобразование Лапласа к левой и правой части получим:

$$
A(s)Y(s)=B(s)X(s).
$$
где:
$$
A(s) = \sum^{n}_{0}a_is^i, \qquad B(s) = \sum^{m}_{0}b_is^i
$$

Передаточная функция системы определяется как отношение изображения выходного сигнала к изображению входного сигнала при нулевых начальных условиях:

$$
H(s)=\frac{Y(s)}{X(s)} = \frac{B(s)}{A(s)}.
$$


###  Полюса и нули передаточной функции

Передаточная функция может быть представлена в виде:
$$
H(s)=\frac{\sum^{m}_{0}{b_i\cdot s^{i}}}{\sum^{n}_{0}{a_i\cdot s^{i}}}=
K\frac{(s-z_1)(s-z_2)\ldots(s-z_m)}
{(s-p_1)(s-p_2)\ldots(s-p_n)},
$$

где $z_i$ — нули, а $p_i$ — полюса передаточной функции, $K$ - коэффициент усиления

Полюса и нули определяются как корни числителя и знаменателя передаточной функции соответственно. Система с передаточной функией $H(s)$ является устойчивой, если полюса находятся в левой полуплоскости: $Re(p_i)<0$.


### 2. Дискретные системы


###  Z-преобразование
Двухстороннее Z-преобразование:
$$
X(z)=\mathcal{Z}\{x[n]\}
=\sum_{n=-\infty}^{\infty}x[n]z^{-n}.
$$

Для дискретных сигналов используется одностороннее Z-преобразование:

$$
X(z)=\mathcal{Z}\{x[n]\}
=\sum_{n=0}^{\infty}x[n]z^{-n}.
$$

Аналогично:

$$
Y(z)=\mathcal{Z}\{y[n]\}.
$$

Для дискретной системы передаточная функция определяется как:

$$
W(z)=\frac{Y(z)}{X(z)}.
$$

Следовательно:

$$
Y(z)=W(z)X(z).
$$

Дискретный выходной сигнал находится с помощью обратного Z-преобразования:

$$
y[n]=\mathcal{Z}^{-1}\{Y(z)\}.
$$

### Билинейное преобразование

Для перехода от передаточной функции непрерывной системы $H(s)$ к передаточной функции дискретной системы $H(z)$ может использоваться билинейное преобразование:

$$
s=\frac{2}{T_s}\frac{z-1}{z+1},
$$

где $T_s$ — период дискретизации.

###  Рекуррентное уравнение

После получения дискретной передаточной функции

$$
W(z)=\frac{Y(z)}{X(z)}
$$

её можно представить в виде:

$$
W(z)=
\frac{b_0+b_1z^{-1}+\ldots+b_mz^{-m}}
{1+a_1z^{-1}+\ldots+a_nz^{-n}}.
$$

Тогда связь между входным и выходным сигналами задаётся рекуррентным уравнением:

$$
y[n]
=
b_0x[n]+b_1x[n-1]+\ldots+b_mx[n-m]
-
a_1y[n-1]-\ldots-a_ny[n-n].
$$

Рекуррентное уравнение позволяет рассчитывать значение выходного сигнала $y[n]$ последовательно для каждого дискретного момента времени.

### 3. Преобразование Фурье непрерывных сигналов

#### 3.1. Определение

Прямое и обратное преобразования Фурье:

$$
X(\omega)=\int_{-\infty}^{\infty}x(t)e^{-j\omega t}\,dt,
\qquad
x(t)=\frac{1}{2\pi}\int_{-\infty}^{\infty}X(j\omega)e^{j\omega t}\,d\omega.
$$

#### 3.1.1. Спектр и спектральная плотность

**Спектр** сигнала показывает, из каких гармонических составляющих он состоит, с какими амплитудами и фазами. **Спектральная плотность** показывает, как энергия или мощность сигнала распределена по частоте.

 Спектр комплексный и позволяет восстановить сигнал, спектральная плотность вещественна и фазу не содержит.

- **Спектр непериодического сигнала.** Для сигнала с конечной энергией спектром называется его преобразование Фурье
$$
X(j\omega)=\int_{-\infty}^{\infty}x(t)e^{-j\omega t}\,dt,
\qquad
x(t)=\frac{1}{2\pi}\int_{-\infty}^{\infty}X(j\omega)e^{j\omega t}\,d\omega.
$$
Модуль $|X(j\omega)|$ называется амплитудным спектром, аргумент $\arg X(j\omega)$ — фазовым спектром. Спектр непрерывен по частоте, а его размерность равна размерности сигнала, умноженной на время (например, В·с).

- **Спектр периодического сигнала.** Для сигнала с периодом $T_0$ ($\omega_0=2\pi/T_0$) используется ряд Фурье:
$$
x(t)=\sum_{k=-\infty}^{\infty}c_ke^{jk\omega_0t}, \qquad
c_k=\frac{1}{T_0}\int_{0}^{T_0}x(t)e^{-jk\omega_0t}\,dt.
$$
Спектр такого сигнала линейчатый: он состоит из отдельных линий на частотах $k\omega_0$, а коэффициенты $c_k$ имеют ту же размерность, что и сам сигнал. Амплитудный спектр задаётся значениями $|c_k|$, фазовый — $\arg c_k$. Периодический сигнал имеет бесконечную энергию, поэтому его преобразование Фурье существует только в обобщённом смысле:
$$
X(j\omega)=2\pi\sum_{k=-\infty}^{\infty}c_k\,\delta(\omega-k\omega_0).
$$

**Спектральная плотность энергии.** Для сигнала с конечной энергией
$$
E_x(\omega)=|X(j\omega)|^2,
\qquad
E=\int_{-\infty}^{\infty}x^2(t)\,dt=\frac{1}{2\pi}\int_{-\infty}^{\infty}E_x(\omega)\,d\omega
$$
(равенство Парсеваля). Величина $E_x(\omega)\,d\omega/2\pi$ равна энергии, заключённой в полосе $d\omega$.


Спектральная плотность $E_x(\omega)$ вещественна и неотрицательна. Фазовой информации не содержит, поэтому по ним сигнал восстановить нельзя.


#### 3.2. Связь с преобразованием Лапласа

Преобразование Фурье — это преобразование Лапласа, вычисленное на мнимой оси $s=j\omega$:

$$
X(j\omega)=X(s)\big|_{s=j\omega},
$$

если мнимая ось входит в область сходимости. Для передаточной функции $H(s)$ это выполняется, когда система устойчива. Тогда

$$
H(j\omega)=\mathcal{F}\{h(t)\}=H(s)\big|_{s=j\omega}
$$

называется частотной характеристикой системы. Если система неустойчива, выражение $H(s)|_{s=j\omega}$ можно вычислить, но оно уже не является фурье-образом импульсной характеристики и не описывает установившийся режим.

#### 3.3. Свойства непрерывного преобразования Фурье
| Свойство | Сигнал | Спектр |
|----|----|----|
| Линейность | $\alpha x_1(t)+\beta x_2(t)$ | $\alpha X_1(j\omega)+\beta X_2(j\omega)$ |
| Сдвиг по времени | $x(t-\tau)$ | $e^{-j\omega\tau}X(j\omega)$ |
| Сдвиг по частоте (модуляция) | $x(t)e^{j\omega_0t}$ | $X(j(\omega-\omega_0))$ |
| Масштабирование | $x(at)$ | $\frac{1}{\lvert a\rvert}X\!\left(j\frac{\omega}{a}\right)$ |
| Дифференцирование | $\frac{dx(t)}{dt}$ | $j\omega X(j\omega)$ |
| Свёртка во времени | $(x*h)(t)$ | $X(j\omega)H(j\omega)$ |
| Умножение во времени | $$x_1(t)x_2(t)$$  | $$\frac{1}{2\pi}(X_1*X_2)(j\omega)$$ |

Теорема Парсеваля:
$$
\int_{-\infty}^{\infty}|x(t)|^2dt=\frac{1}{2\pi}\int_{-\infty}^{\infty}|X(j\omega)|^2d\omega.
$$


### 4. Дискретное преобразование Фурье

#### 4.1. Определение

Для конечной последовательности $x[n]$, $n=0,\ldots,N-1$:

$$
X[k]=\sum_{n=0}^{N-1}x[n]e^{-j\frac{2\pi}{N}kn},
\qquad
x[n]=\frac{1}{N}\sum_{k=0}^{N-1}X[k]e^{j\frac{2\pi}{N}kn}.
$$


$$
X[k]=X(e^{j\omega})\Big|_{\omega=2\pi k/N},
\qquad
f_k=k\,\frac{f_s}{N}.
$$

Расстояние между соседними отсчётами спектра (шаг по частоте) равно $\Delta f=\dfrac{f_s}{N}=\dfrac{1}{NT_s}$.

### 4.2. Связь с $Z$-преобразованием

Для дискретной системы с передаточной функцией $H(z)$ частотная характеристика получается подстановкой $z=e^{j\omega}$, то есть вычислением $H(z)$ на единичной окружности:

$$
H(e^{j\omega})=H(z)\big|_{z=e^{j\omega}}=\sum_{n=0}^{\infty}h[n]e^{-j\omega n},
\qquad
\omega=\frac{2\pi f}{f_s}.
$$


Функция $H(e^{j\omega})$ периодична по $\omega$ с периодом $2\pi$, а при вещественных коэффициентах $H(e^{-j\omega})=H^*(e^{j\omega})$. Поэтому её строят на $\omega\in[0,\pi]$ (иначе $f\in[0,1]$), что соответствует $f\in[0,f_s/2]$.


### 4.3. Свойства

| Свойство | Сигнал | Спектр |
|---|---|---|
| Линейность | $\alpha x_1[n]+\beta x_2[n]$ | $\alpha X_1+\beta X_2$ |
| Сдвиг по времени | $x[n-n_0]$ | $e^{-j\omega n_0}X(e^{j\omega})$ |
| Модуляция | $x[n]e^{j\omega_0n}$ | $X(e^{j(\omega-\omega_0)})$ |
| Свёртка | $(x*h)[n]$ | $X(e^{j\omega})\,H(e^{j\omega})$ |
| Умножение | $x_1[n]x_2[n]$ | $\frac{1}{2\pi}\int_{-\pi}^{\pi}X_1(e^{j\theta})X_2(e^{j(\omega-\theta)})d\theta$ |
| Дифференцирование по $\omega$ | $$n\,x[n]$$ | $$j\,\dfrac{dX(e^{j\omega})}{d\omega}$$ |




$$
\sum_{n=-\infty}^{\infty}|x[n]|^2=\frac{1}{2\pi}\int_{-\pi}^{\pi}|X(e^{j\omega})|^2d\omega.
$$


### 4.4. Спектр

Для вещественной последовательности $x[n]$ выполняется

$$
X[N-k]=X^*[k]\ \Rightarrow\ |X[N-k]|=|X[k]|,
$$

то есть спектр симметричен относительно $f_s/2$. Информативны только отсчёты $k=0,\ldots,\lfloor N/2\rfloor$, которым соответствуют частоты $f_k=k\,f_s/N\in[0,f_s/2]$. Отсчёты с $k>N/2$ соответствуют отрицательным частотам $f_k-f_s$. Для комплексной последовательности симметрии нет, и нужны все $N$ отсчётов.

#### 4.4.1 Типы спектров

Дискретность в одной области приводит к периодичности в другой. Отсюда четыре случая:

- непериодический непрерывный сигнал: спектр сплошной (спектральная плотность), непериодический по частоте;
- периодический непрерывный сигнал: спектр линейчатый (ряд Фурье), непериодический по частоте;
- дискретный непериодический сигнал: спектр сплошной, периодический по частоте с периодом $f_s$;
- дискретный периодический сигнал (или конечная последовательность, которую ДПФ считает периодической с периодом $N$): спектр линейчатый и периодический, то есть $X[k+N]=X[k]$. Именно этот случай описывает ДПФ.

Периодичность спектра дискретного сигнала — следствие дискретизации во времени: копии спектра повторяются с шагом $f_s$. Если они перекрываются (в сигнале есть компоненты выше $f_s/2$), возникает алиасинг.


#### 4.4.2. Масштабирование амплитудного спектра

Если сигнал $x[n]=A\sin(2\pi f_0nT_s)$ содержит целое число периодов на длине записи (то есть $f_0=k_0f_s/N$), то

$$
|X[k_0]|=\frac{AN}{2},\qquad\text{откуда}\qquad A=\frac{2|X[k_0]|}{N}.
$$

Для постоянной составляющей $A_0=|X[0]|/N$. Для оконных функций нормировка делается на сумму отсчётов окна: $A\approx 2|X[k_0]|/\sum_n w[n]$.

#### 4.4.3. Спектр и спектральная плотность

| Признак | Спектр | Спектральная плотность |
|---|---|---|
| **Тип сигнала** | Периодический | Непериодический или случайный |
| **Вид спектра** | Дискретный (линейчатый) | Непрерывный (сплошной) |
| **Физический смысл** | Амплитуды и фазы конкретных гармоник | Плотность распределения энергии/мощности на 1 Гц |
| **Единицы измерения** | В единицах сигнала (В, А и т.д.) | В единицах сигнала, делённых на Гц (В/Гц, Вт/Гц) |
