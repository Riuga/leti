## **Теоретическая часть**

### 1. Синтез КИХ-фильтров

#### 1.1. Способы синтеза КИХ-фильтров

**Метод проектирования с помощью окон**

Проектирование КИХ-фильтров методом окон начинается с выбора требуемой частотной характеристики.

Для определённости рассмотрим фильтр нижних частот (ФНЧ). В качестве исходной принимается идеальная частотная характеристика, равная единице в полосе пропускания $0\le f\le f_c$ и нулю на частотах, превышающих частоту среза $f_c$:

$$
H(f)=\begin{cases}1, & 0\le f\le f_c,\\ 0, & f_c<f\le f_s/2.\end{cases}
$$

Дискретным представлением такого фильтра является периодическая частотная характеристика с периодом, равным частоте дискретизации $f_s$. Поэтому далее все частотные характеристики рассматриваются на интервале $0\ldots f_s$. Для вещественного фильтра они симметричны относительно $f_s/2$, и на графиках обычно изображается только интервал $0\ldots f_s/2$.

Для того чтобы получить коэффициенты фильтра $h(n)$ по заданной частотной характеристике, необходимо выполнить обратное преобразование Фурье. Это можно сделать двумя способами.

**Аналитический способ.** Согласно определению обратного преобразования Фурье для дискретных сигналов, импульсная характеристика связана с частотной характеристикой соотношением

$$
h(n)=\frac{1}{2\pi}\int_{-\pi}^{\pi}H(\omega)\,e^{j\omega n}\,d\omega,\qquad \omega=\frac{2\pi f}{f_s}.
$$

Порядок действий следующий:

1. Записать выражение для дискретной частотной характеристики $H(\omega)$: она равна единице при $|\omega|\le\omega_c$ и нулю при $\omega_c<|\omega|\le\pi$, где $\omega_c=2\pi f_c/f_s$.
2. Подставить это выражение в формулу обратного преобразования Фурье, при этом пределы интегрирования сужаются до полосы пропускания:
$$
h(n)=\frac{1}{2\pi}\int_{-\omega_c}^{\omega_c}e^{j\omega n}\,d\omega .
$$
3. Вычислить интеграл и получить выражение для $h$ как функции индекса времени $n$:
$$
h(n)=\frac{\sin(\omega_c n)}{\pi n},\qquad h(0)=\frac{\omega_c}{\pi}=\frac{2f_c}{f_s}.
$$

**Численный способ.** Частотная характеристика задаётся $N$ отсчётами $H(m)$, $m=0,\ldots,N-1$, на интервале $0\ldots f_s$ с учётом симметрии относительно $f_s/2$. Коэффициенты определяются обратным дискретным преобразованием Фурье:

$$
h(k)=\frac{1}{N}\sum_{m=0}^{N-1}H(m)\,e^{\,j2\pi mk/N},\qquad k=0,\ldots,N-1.
$$

**Приведение к реализуемому виду.** Полученная в аналитическом способе последовательность $h(n)$ бесконечна, поэтому реализовать фильтр с такой характеристикой невозможно. Для того чтобы сделать фильтр реализуемым, необходимо выполнить два действия.

1. *Ограничить длину последовательности* $N$ отсчётами (число $N$ выбирается нечётным) и сдвинуть её вправо на $M=(N-1)/2$ отсчётов
$$
h_{\text{id}}(n)=\frac{\sin\big(\omega_c(n-M)\big)}{\pi(n-M)},\quad n\ne M;\qquad h_{\text{id}}(M)=\frac{\omega_c}{\pi},\qquad n=0,\ldots,N-1.
$$

Сдвиг на $M=(N-1)/2$ переносит последовательность в диапазон $n=0,\ldots,N-1$: $$n\in[-M,\,M]\ \longrightarrow\ n\in[0,\,N-1].$$


2. *Умножить результат на оконную функцию* $w(n)$:
$$
h(n)=h_{\text{id}}(n)\,w(n).
$$

 Простое усечение импульсной характеристики эквивалентно умножению на прямоугольное окно. В силу свойства преобразования Фурье умножению во временной области соответствует свёртка в частотной, поэтому реальная АЧХ представляет собой свёртку идеальной характеристики со спектром окна. Это приводит к пульсациям АЧХ в полосах пропускания и подавления (явление Гиббса, выброс около 9 %), которые не уменьшаются с ростом $N$. Применение окон, плавно спадающих к краям, уменьшает уровень боковых лепестков их спектра и, как следствие, пульсации, однако расширяет главный лепесток и тем самым переходную область между полосами.



#### 1.2. Синтез фильтра нижних частот

Исходными данными являются частота дискретизации $f_s$, частота среза $f_c$, тип окна и нечётная длина фильтра $N$. Для того чтобы синтезировать ФНЧ, необходимо выполнить следующие действия.

1. Определить нормированную частоту среза $\omega_c=2\pi f_c/f_s$ и задержку $M=(N-1)/2$.
2. Рассчитать коэффициенты идеального фильтра $h_{\text{id}}(n)$ по формуле, полученной в п. 1.1.

3. Умножить их на выбранную оконную функцию:
$$
h_{\text{lp}}(n)=h_{\text{id}}(n)\,w(n).
$$
4. Выполнить нормировку, чтобы коэффициент передачи на нулевой частоте был равен единице:
$$
h_{\text{lp}}(n)\leftarrow\frac{h_{\text{lp}}(n)}{\sum_{k=0}^{N-1}h_{\text{lp}}(k)}.
$$
5. Для проверки рассчитать АЧХ как ДПФ коэффициентов, дополненных нулями.

Так как коэффициенты симметричны, $h(n)=h(N-1-n)$, синтезированный фильтр обладает линейной ФЧХ и постоянной групповой задержкой, равной $M$ отсчётам.

#### 1.3. Синтез полосового фильтра и фильтра верхних частот

Полосовой фильтр и фильтр верхних частот могут быть получены из ФНЧ-прототипа путём сдвига его частотной характеристики по оси частот. Согласно свойству сдвига спектра, умножение импульсной характеристики на комплексную экспоненту переносит спектр на частоту $\omega_0$:

$$
h(n)\,e^{\,j\omega_0 n}\ \longleftrightarrow\ H(\omega-\omega_0).
$$

Поскольку коэффициенты реального фильтра должны быть вещественными, а его АЧХ на интервале $0\ldots f_s$ симметрична относительно $f_s/2$, сдвигаются одновременно две симметричные копии спектра. Для этого используется множитель $2\cos(\omega_0 n)$:

$$
2h(n)\cos(\omega_0 n)\ \longleftrightarrow\ H(\omega-\omega_0)+H(\omega+\omega_0).
$$

**Фильтр верхних частот.** ФВЧ получается сдвигом спектра ФНЧ на $f_s/2$ ($\omega_0=\pi$). Для того чтобы получить ФВЧ с частотой среза $f_c$, необходимо:

1. Определить частоту среза прототипа $f_c'=f_s/2-f_c$.
2. Синтезировать ФНЧ с частотой среза $f_c'$ по алгоритму п. 1.2.
3. Выполнить сдвиг спектра, учитывая, что $e^{j\pi n}=(-1)^n$ (свойство сдвига по частоте):
$$
h_{\text{hp}}(n)=(-1)^{\,n-M}\,h_{\text{lp}}(n).
$$

При $\omega_0=\pi$ обе копии спектра совпадают, поэтому множитель 2 в данном случае не используется.

**Полосовой фильтр.** Пусть заданы границы полосы пропускания $f_1$ и $f_2$. Для того чтобы синтезировать полосовой фильтр, необходимо:

1. Определить центральную частоту и полуширину полосы:
$$
f_0=\frac{f_1+f_2}{2},\qquad f_c'=\frac{f_2-f_1}{2}.
$$
2. Синтезировать ФНЧ-прототип с частотой среза $f_c'$ по алгоритму п. 1.2.
3. Сдвинуть его спектр на $\omega_0=2\pi f_0/f_s$, умножив импульсную характеристику на косинус, отсчитываемый от центра фильтра:
$$
h_{\text{bp}}(n)=2\,h_{\text{lp}}(n)\cos\big(\omega_0(n-M)\big).
$$

Множитель 2 компенсирует деление амплитуды между двумя копиями спектра, расположенными вблизи $f_0$ и $f_s-f_0$. Отсчёт аргумента от $n-M$ сохраняет симметрию коэффициентов и, следовательно, линейность ФЧХ. Условием применимости метода является $0<f_1<f_2<f_s/2$.

### 2. Групповая задержка
Групповая задержка системы определяется выражением
$$
\tau_g(\omega)=-\frac{d\varphi(\omega)}{d\omega},
$$
где $\varphi(\omega)$ — ФЧХ. Она измеряется в отсчётах; для перехода к секундам нужно разделить её на $f_s$. Задержка, с которой передаётся гармоническое колебание, определяется *фазовой* задержкой $\tau_p(\omega)=-\varphi(\omega)/\omega$. Задержка огибающей узкополосного сигнала определяется групповой задержкой.
